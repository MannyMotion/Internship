```python
import pandas as pd
import os

# Load each table
dim_clinic = pd.read_csv(r'C:/Users/Invate/Downloads/DIM_CLINICS.csv')
dim_date = pd.read_csv(r'C:\Users\Invate\Downloads\DIM_DATES.csv')
dim_patient = pd.read_csv(r'C:\Users\Invate\Downloads\DIM_PATIENT.csv')
dim_provider = pd.read_csv(r'C:\Users\Invate\Downloads\DIM_PROVIDERS.csv')
fact_appointment = pd.read_csv(r'C:\Users\Invate\Downloads\FACT_APPOINTMENTS.csv')

# Merge step 1: Merge fact_appointment with dim_patient
merged = fact_appointment.merge(dim_patient, on='PATIENT_ID', how='left')

# Step 2: Merge with dim_provider
merged = merged.merge(dim_provider, on='PROVIDER_ID', how='left')

# Step 3: Merge with dim_clinic
merged = merged.merge(dim_clinic, on='PROVIDER_CLINIC_ID', how='left')

# Step 4: Merge with dim_date
merged = merged.merge(dim_date, on='DATE_ID', how='left')

# Save the final merged dataset
merged.to_csv('merged_healthpulse_data.csv', index=False)

print("Merge complete! The file is saved as 'merged_healthpulse_data.csv'.")
```
